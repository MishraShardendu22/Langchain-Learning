```python
# build a streaming RAG chatbot

def streaming_rag(query: str):
    retrieved_docs = retriever.invoke(query)

    context = "\n\n".join(
        [doc.page_content for doc in retrieved_docs]
    )

    prompt = f"""
Use the provided context to answer the question.

Context:
{context}

Question:
{query}

Answer:
"""

    response = llm.stream(prompt)

    full_response = ""

    for chunk in response:
        print(chunk.content, end="", flush=True)
        full_response += chunk.content

    return full_response


streaming_rag(
    "Explain encoder-decoder architecture in Transformers"
)
```

```python
query = "Why is self-attention better than recurrence?"

initial_docs = retriever.invoke(query)

print(f"Initial Retrieved Docs: {len(initial_docs)}")

query_words = set(
    query.lower().split()
)

scored_docs = []

for doc in initial_docs:
    content_words = set(
        doc.page_content.lower().split()
    )

    score = len(
        query_words.intersection(content_words)
    )

    scored_docs.append((score, doc))

reranked_docs = sorted(
    scored_docs,
    key=lambda x: x[0],
    reverse=True
)

top_docs = [
    doc for score, doc in reranked_docs[:3]
]

print("\nTOP RERANKED DOCS:\n")

for i, doc in enumerate(top_docs, start=1):
    print("\n" + "=" * 80)
    print(f"RERANKED DOC {i}")
    print(doc.page_content[:1200])
```

```python
def generate_queries(user_query: str):
    prompt = f"""
Generate 5 different search queries
for the following user question.

Question:
{user_query}

Only return the queries.
"""

    response = llm.invoke(prompt)

    queries = response.content.split("\n")

    clean_queries = [
        q.strip("- ").strip()
        for q in queries
        if q.strip()
    ]

    return clean_queries


queries = generate_queries(
    "Why are Transformers better than RNNs?"
)

print("Generated Queries:\n")

for q in queries:
    print(q)
```

```python
import os

from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

FAISS_DB_PATH = "faiss_index"

if os.path.exists(FAISS_DB_PATH):

    print("Loading Existing FAISS Index...")

    vectorstore = FAISS.load_local(
        FAISS_DB_PATH,
        embedding_model,
        allow_dangerous_deserialization=True
    )

else:

    print("Creating New FAISS Index...")
    loader = PyMuPDFLoader(
        "attention.pdf"
    )

    documents = loader.load()

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=50
    )

    chunks = text_splitter.split_documents(documents)

    vectorstore = FAISS.from_documents(
        chunks,
        embedding_model
    )

    vectorstore.save_local(
        FAISS_DB_PATH
    )

    print("FAISS Index Saved")

retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

query = "What is multi head attention?"

results = retriever.invoke(query)

print("\nRETRIEVED RESULTS:\n")

for idx, doc in enumerate(results, start=1):

    print(f"RESULT {idx}")
    print(doc.page_content[:400])
    print("-" * 50)
```

```python
documents = [
    Document(
        page_content="""
        Kubernetes was created by Google engineers.
        """
    ),

    Document(
        page_content="""
        FAISS is a vector similarity search library developed by Meta.
        """
    )
]

dataset = [
    {
        "question": "Who created Kubernetes?",
        "answer": "Google engineers"
    },

    {
        "question": "What is FAISS?",
        "answer": "A vector similarity search library developed by Meta"
    }
]
```

```python
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = FAISS.from_documents(
    documents,
    embedding_model
)

retriever = vectorstore.as_retriever()
```

```python
def rag_app(inputs):

    question = inputs["question"]

    retrieved_docs = retriever.invoke(
        question
    )

    context = "\n\n".join(
        [doc.page_content for doc in retrieved_docs]
    )

    prompt = f"""
    Answer ONLY using the provided context.

    Context:
    {context}

    Question:
    {question}
    """

    response = llm.invoke(prompt)

    return {
        "answer": response.content,
        "documents": retrieved_docs
    }
```

```python
def correctness_evaluator(
    inputs,
    outputs,
    reference_outputs
):
    question = inputs["question"]

    predicted_answer = outputs["answer"]

    reference_answer = reference_outputs["answer"]

    prompt = f"""
    Question:
    {question}

    Reference Answer:
    {reference_answer}

    Model Answer:
    {predicted_answer}

    Is the model answer correct?

    Respond only YES or NO.
    """

    response = llm.invoke(prompt)

    score = (
        "YES" in response.content.upper()
    )

    return {
        "key": "correctness",
        "score": score
    }
```
